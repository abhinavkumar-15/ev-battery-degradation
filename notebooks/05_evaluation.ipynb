# 05 – Evaluation
Reads metrics written by the pipeline (no retraining).

```python
import sys, warnings; sys.path.insert(0, '..'); warnings.filterwarnings('ignore')
import pandas as pd, numpy as np, matplotlib.pyplot as plt
DATASET = 'demo'   # change to 'nasa' once data/raw/nasa is populated
import json
m = json.load(open(f'../outputs/metrics/{DATASET}/forecast_h10_metrics.json'))['metrics']
pd.DataFrame({k:{x:v[x] for x in ('mae','rmse','r2','r2_delta')} for k,v in m.items()}).T
```

```python
p = pd.read_csv(f'../outputs/predictions/{DATASET}_forecast_h10.csv')
best = json.load(open(f'../outputs/metrics/{DATASET}/forecast_h10_metrics.json'))['best_model']
res = p.y_true - p[best]
print(best, 'residual mean %.3f std %.3f' % (res.mean(), res.std()))
p.assign(res=res).groupby('battery_id').res.agg(['mean','std'])
```
