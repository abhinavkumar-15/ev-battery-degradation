# 01 – Data exploration
Loads the cycle table through the dataset adapter and inspects capacity fade. **Demo data is synthetic.**

```python
import sys, warnings; sys.path.insert(0, '..'); warnings.filterwarnings('ignore')
import pandas as pd, numpy as np, matplotlib.pyplot as plt
DATASET = 'demo'   # change to 'nasa' once data/raw/nasa is populated
from src.data.adapters import ADAPTERS
raw = ADAPTERS[DATASET]().load()
print(raw.shape); raw.head()
```

```python
raw.groupby('battery_id').agg(cycles=('cycle','max'), cap0=('capacity_ah','first'), capN=('capacity_ah','last'))
```

```python
fig, ax = plt.subplots(figsize=(7,4))
for b,g in raw.groupby('battery_id'): ax.plot(g.cycle, g.capacity_ah, label=b)
ax.set(xlabel='cycle', ylabel='capacity (Ah)'); ax.legend(ncol=2, fontsize=7); plt.show()
```
